#### 2.1.2 Filling Missing Values

```python
df_cleaned = df.copy()

numeric_columns = df_cleaned.select_dtypes(include=np.number).columns

for col in numeric_columns:

    df_cleaned[col] = df_cleaned[col].fillna(df_cleaned[col].median())

print("Missing values filled using Median")
```

#### 2.1.3 Filling Missing Values Using Mode

```python
categorical_columns = df_cleaned.select_dtypes(include=['object']).columns

for col in categorical_columns:

    if df_cleaned[col].isnull().sum() > 0:

        df_cleaned[col] = df_cleaned[col].fillna(df_cleaned[col].mode()[0])

print("Missing values filled using Mode")
```

#### Hashtag Count Extraction

```python
hashtag_col = next(
    (col for col in df_cleaned.columns if 'hashtag' in col.lower()),
    None
)

if hashtag_col:
    df_cleaned['hashtag_count'] = (
        df_cleaned[hashtag_col].fillna('').astype(str).str.count('#')
    )
    print(df_cleaned[['hashtag_count']].head())
else:
    print("Hashtag column not found. Check the dataset columns.")
```

#### Cleaning Sentiment Labels

```python
sentiment_col = next(
    (col for col in df_cleaned.columns if 'sentiment' in col.lower()),
    None
)

if sentiment_col:
    df_cleaned[sentiment_col] = (
        df_cleaned[sentiment_col].astype('string').str.strip().str.title()
    )
    print(df_cleaned[sentiment_col].value_counts(dropna=False))
else:
    print("Sentiment column not found.")
```

## Task 4: Data Wrangling

```python
# Identify likes, comments and shares columns
likes_col = next((c for c in df_cleaned.columns if 'like' in c.lower()), None)
comments_col = next((c for c in df_cleaned.columns if 'comment' in c.lower()), None)
shares_col = next((c for c in df_cleaned.columns if 'share' in c.lower()), None)

if all([likes_col, comments_col, shares_col]):
    df_cleaned['engagement_score'] = (
        df_cleaned[likes_col].fillna(0) +
        df_cleaned[comments_col].fillna(0) +
        df_cleaned[shares_col].fillna(0)
    )
    print("Engagement Score Created Successfully!")
    display(df_cleaned.head())
else:
    print("Likes, Comments or Shares column not found.")
```

### Log Transformation

```python
import numpy as np

if likes_col:
    df_cleaned['log_likes'] = np.log1p(df_cleaned[likes_col])

if comments_col:
    df_cleaned['log_comments'] = np.log1p(df_cleaned[comments_col])

if shares_col:
    df_cleaned['log_shares'] = np.log1p(df_cleaned[shares_col])

print("Log Metrics Created Successfully!")
```

### GroupBy Summary Analysis

```python
group_columns = ['post_type', 'country', 'sentiment']

for col in group_columns:
    actual_col = next(
        (c for c in df_cleaned.columns if c.lower().replace(' ', '_') == col),
        None
    )

    if actual_col and 'engagement_score' in df_cleaned.columns:
        print("\nEngagement by", actual_col)

        display(
            df_cleaned.groupby(actual_col)['engagement_score']
            .mean()
            .reset_index()
        )
```

### 6.1 Matplotlib Visualizations
#### 1. Scatter Plot – Likes vs Impressions

```python
likes_col = next(
    (c for c in df_cleaned.columns if 'like' in c.lower()), None
)

impressions_col = next(
    (c for c in df_cleaned.columns if 'impression' in c.lower()), None
)

if likes_col and impressions_col:
    plt.figure(figsize=(8, 5))

    plt.scatter(
        df_cleaned[impressions_col],
        df_cleaned[likes_col]
    )

    plt.title("Likes vs Impressions")
    plt.xlabel("Impressions")
    plt.ylabel("Likes")
    plt.show()
else:
    print("Likes or Impressions column not found.")
```

#### 2. Line Chart – Daily Engagement Trend

```python
date_col = 'posted_at'

df_cleaned['posted_at'] = pd.to_datetime(
    df_cleaned['posted_at'],
    errors='coerce'
)

print(df_cleaned['posted_at'].dtype)
```

```python
daily_engagement = (
    df_cleaned.groupby(df_cleaned['posted_at'].dt.date)
    ['engagement_score']
    .mean()
    .reset_index()
)

plt.figure(figsize=(12, 5))

plt.plot(
    daily_engagement['posted_at'],
    daily_engagement['engagement_score'],
    marker='o'
)

plt.title("Daily Engagement Trend")
plt.xlabel("Date")
plt.ylabel("Average Engagement Score")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
